```python
class BankAccount:
    """Models a simple bank account using OOP concepts."""

    def __init__(self, holder, balance):
        self.holder = holder
        self.__balance = balance

    def deposit(self, amount):
        """Add money to the balance. Rejects negative or zero amounts."""
        if amount <= 0:
            print(f"Deposit failed: amount must be greater than 0.")
            return
        self.__balance += amount
        print(f"Deposit of ${amount:,.2f} successful for {self.holder}.")

    def withdraw(self, amount):
        """Subtract money from the balance. Rejects amounts greater than the balance."""
        if amount <= 0:
            print("Withdrawal failed: amount must be greater than 0.")
            return
        if amount > self.__balance:
            print(f"Withdrawal failed: insufficient funds for {self.holder}.")
            return
        self.__balance -= amount
        print(f"Withdrawal of ${amount:,.2f} successful for {self.holder}.")

    def check_balance(self):
        """Display the current balance."""
        print(f"{self.holder}'s current balance: ${self.__balance:,.2f}")



account1 = BankAccount("Raúl Pérez", 5000)
account2 = BankAccount("Rodrigo Torres", 3000)



print("----- Raúl Pérez's Account -----")
print(f"Holder: {account1.holder}")
account1.check_balance()
account1.deposit(1500)
account1.withdraw(2000)
account1.check_balance()

print("\n----- Rodrigo Torres's Account -----")
print(f"Holder: {account2.holder}")
account2.check_balance()
account2.deposit(750)
account2.withdraw(4000)
account2.check_balance()
```
